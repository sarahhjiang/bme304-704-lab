### ⭐ 1.2: Bonus - Linear Algebra Notation for Multiple Linear Regression

We can rewrite the OLS equations above like this:

\begin{align}
  \hat Y &= \begin{pmatrix}  
             1 & x_1 \\
             1 & x_2 \\
             \vdots & \vdots \\
             1 & x_n
          \end{pmatrix} \begin{pmatrix}
                               \beta_0 \\
                               \beta_1
                              \end{pmatrix} \\ 
         &= \begin{pmatrix}  
             1 & \mid \\
             \vdots & \vec{x} \\
             1 & \mid
          \end{pmatrix} \begin{pmatrix}
                               \beta_0 \\
                               \beta_1
                              \end{pmatrix} \\ 
         &= X \vec{\beta}
\end{align}
